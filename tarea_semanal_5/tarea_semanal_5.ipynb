## Circuito T

Para comenzar a resolver la Matriz Admitancia Indefinida (MAI) debo definir un nodo intermedio extra que llamaré $V4$:

<div style="text-align: center;">
  <img src="recursos/circuito_t_nodos.png" width="300">
</div>

Ahora planteo las ecuaciones de nodos en cada lugar:

1) $I1 = V1(\frac{1}{SL}) - V4(\frac{1}{SL})$
2) $I2 = V2(\frac{1}{SL}) - V4(\frac{1}{SL})$
3) $I3 = V3(SC) - V4(\frac{1}{SL})$
4) $I4 = V4(\frac{1}{SL} + \frac{1}{SL} + SC) - V1(\frac{1}{SL}) - V2(\frac{1}{SL} - V3(SC) - V4(\frac{1}{SL})$

Para la obtención de cada una de las admitancias partiendo del planteo de nodos uso la siguiente expresión:

$$\large
y_{ij} = \left.\frac{I_i}{V_j}\right|_{V_k=0, \forall_k \ne j}
$$

### Obtención de la MAI

#### Fila 1

$$Y_{11} = \left.\frac{I_1}{V_1}\right|_{V_2=V_3=V_4=0} = \frac{1}{SL}$$

$$Y_{12} = \left.\frac{I_1}{V_2}\right|_{V_1=V_3=V_4=0} = 0$$

$$Y_{13} = \left.\frac{I_1}{V_3}\right|_{V_1=V_2=V_4=0} = 0$$

$$Y_{14} = \left.\frac{I_1}{V_4}\right|_{V_1=V_2=V_3=0} = -\frac{1}{SL}$$

#### Fila 2

$$Y_{21} = \left.\frac{I_2}{V_1}\right|_{V_2=V_3=V_4=0} = 0$$

$$Y_{22} = \left.\frac{I_2}{V_2}\right|_{V_1=V_3=V_4=0} = \frac{1}{SL}$$

$$Y_{23} = \left.\frac{I_2}{V_3}\right|_{V_1=V_2=V_4=0} = 0$$

$$Y_{24} = \left.\frac{I_2}{V_4}\right|_{V_1=V_2=V_3=0} = -\frac{1}{SL}$$

#### Fila 3

$$Y_{31} = \left.\frac{I_3}{V_1}\right|_{V_2=V_3=V_4=0} = 0$$

$$Y_{32} = \left.\frac{I_3}{V_2}\right|_{V_1=V_3=V_4=0} = 0$$

$$Y_{33} = \left.\frac{I_3}{V_3}\right|_{V_1=V_2=V_4=0} = SC$$

$$Y_{34} = \left.\frac{I_3}{V_4}\right|_{V_1=V_2=V_3=0} = -SC$$

#### Fila 4

$$Y_{41} = \left.\frac{I_4}{V_1}\right|_{V_2=V_3=V_4=0} = -\frac{1}{SL}$$

$$Y_{42} = \left.\frac{I_4}{V_2}\right|_{V_1=V_3=V_4=0} = -\frac{1}{SL}$$

$$Y_{43} = \left.\frac{I_4}{V_3}\right|_{V_1=V_2=V_4=0} = -SC$$

$$Y_{44} = \left.\frac{I_4}{V_4}\right|_{V_1=V_2=V_3=0} = \frac{2}{SL}+SC$$

Lo llevo al formato de matriz 4x4:
$$

$$A^{ij}_{mn} = \frac{V_{ij}}{V_{mn}} = sgn(m - n) sgn(i - j) \frac{\underline{Y}^{mn}_{ij}}{\underline{Y}^{mn}_{mn}}$$

$$\frac{V23}{V13} = A^{23}_{13} = sgn(1-3)sgn(2-3)\frac{\underline{Y}^{13}_{23}}{\underline{Y}^{13}_{13}}$$

Con el signo quedando positivo se puede reducir a:

$$A^{23}_{13} = \frac{\underline{Y}^{13}_{23}}{\underline{Y}^{13}_{13}}$$

Por otra definición del mismo capítulo del libro puedo obtener cada cofactor de la siguiente manera:

$$\large \underline{Y}_{ij}^{mn} = (-1)^{(i+j+m+n)} \cdot Y_{ij}^{mn}$$

$$\Delta{Y_{23}^{13}} = \frac{1}{S^2L^2}$$

$$Y_{23}^{13} = \frac{1}{S^2L^2}$$

### Gráficos de las transferencias

Se adjunta el código para la obtención de de los módulos y fases de las tres transferencias. Para las pruebas se asumen valores de $L=10mHy$ y $C=100 \eta F$ así se obtiene una ejemplo de transferencia.

```python
import numpy as np
import matplotlib.pyplot as plt

from pytc2.general import s


L = 10e-3
C = 100e-9

w0 = 1 / np.sqrt(L*C)

transferencias = [
    ((1/(L*C)) / (s**2 + 1/(L*C)), r'$V_{23}/V_{13}$'),
    (s**2 / (s**2 + 1/(L*C)), r'$V_{21}/V_{31}$'),
    (s**2 / (s**2 + 1/(L*C)), r'$V_{12}/V_{32}$')
]

f = np.logspace(1, 6, 2000)
w = 2*np.pi*f

for T, nombre in transferencias:

    H = np.array([complex(T.subs(s, 1j*wi)) for wi in w])

    for valores, ylabel, titulo in [
        (np.abs(H), r'$|T|$', 'Módulo'),
        (np.angle(H, deg=True), 'Fase [°]', 'Fase')
    ]:

        plt.figure(figsize=(10, 6))
        plt.semilogx(f, valores)
        plt.axvline(w0/(2*np.pi), linestyle=':', label=r'$f_0$')
        plt.xlabel('Frecuencia [Hz]')
        plt.ylabel(ylabel)
        plt.title(f'{titulo} de {nombre}')
        plt.grid(True, which='both')
        plt.legend()
        plt.show()
```

### Análisis de los módulos

Analizo dos transferencias:

$$T_T = \frac{V23}{V13} + \frac{V21}{V31}$$

$$\large T_T = \frac{\frac{1}{LC}}{S^2 + \frac{1}{LC}} + \frac{S^2}{S^2 + \frac{1}{LC}}$$

$$\large T_T =\frac{\cancel{S^2 + \frac{1}{LC}}}{\cancel{S^2 + \frac{1}{LC}}}$$

$$\large T_T = 1 \Longrightarrow |T_T| = 1$$

Con eso se demuestra que esas dos transferencias son complementarias.

#### Cálculo de las transferencias

En el siguiente script se calcularán las transferencias $V23/V13$, $V21/V31$ y $V12/V32$.

```python
import sympy as sp

from pytc2.general import s, print_latex, a_equal_b_latex_s
from pytc2.cuadripolos import calc_MAI_vtransf_ij_mn


# Simbolos

L, C = sp.symbols('L C', positive=True)

YL = 1/(s*L)
YC = s*C


# Circuito
#
#    0 -- L ----- 3 ----- L ----- 1
#                 |
#                 C
#                 |
#                 2

# Armo la MAI
Ymai = sp.Matrix([
    [ YL,       0,       0,      -YL],
    [  0,       YL,       0,      -YL],
    [  0,        0,      YC,      -YC],
    [-YL,      -YL,      -YC,  2*YL + YC]
])


# V23/V13
V23_V13 = calc_MAI_vtransf_ij_mn(
    Ymai,
    1, 2,       # V23
    0, 2,       # V13
)

# V21/V31
V21_V31 = calc_MAI_vtransf_ij_mn(
    Ymai,
    1, 0,       # V21
    2, 0,       # V31
)

# V12/V32
V12_V32 = calc_MAI_vtransf_ij_mn(
    Ymai,
    0, 1,       # V12
    2, 1,       # V32
)

print_latex(
    a_equal_b_latex_s(r'\frac{V_{23}}{V_{13}}', V23_V13)
)

print_latex(
    a_equal_b_latex_s(r'\frac{V_{21}}{V_{31}}', V21_V31)
)

print_latex(
    a_equal_b_latex_s(r'\frac{V_{12}}{V_{32}}', V12_V32)
)
```

### Obtención de la MAD

Para obtener la Matriz Admitancia Definida resolveré por el método de nodos planteando conexión en source común.

1) $\large I1 = V1 (S.C_{gs} + S.C_{gd}) - V2.C_{gds}$
2) $I2 = V2. (\frac{1}{r_{ds}} + S.C_{ds} +S.C_{gd}) - V1.(S.C_{gd} - gm)$

Evalúo cada caso pasivando el generador que no estoy analizando.

$$Y_{11} = \left.\frac{I1}{V1}\right|_{V2 = 0} = S(C_{gs} + C_{gd})$$

$$Y_{12} = \left.\frac{I1}{V2}\right|_{V1 = 0} = -S.C_{gd}$$

$$Y_{21} = \left.\frac{I2}{V1}\right|_{V2 = 0} = gm -S.C_{gd}$$

$$Y_{22} = \left.\frac{I2}{V2}\right|_{V1 = 0} = \frac{1}{rds} + S.(C_{ds} + C_{gd})$$

Con esto formo la matriz.

$$Y_{MAD} = \begin{bmatrix}
S(C_{gs} + C_{gd}) & -S.C_{gd} \\
gm -S.C_{gd} &
\frac{1}{rds} + S.(C_{ds} + C_{gd})
\end{bmatrix}$$

### Obtención de la MAI

Ahora con esta matriz puedo hallar la MAI. Para eso tengo que desacoplar el Source.

$$
\begin{pmatrix}
Y_{11} & Y_{12} & Y_{13} \\
Y_{21} & Y_{22} & Y_{23} \\
Y_{31} & Y_{32} & Y_{33} 
\end{pmatrix}
$$

Por propiedad de las MAIs *la sumatoria de las filas es igual a 0* y *la sumatoria de las columnas es igual a 0*

Entonces:
* $Y_{11} + Y_{21} + Y_{31} = 0$
* $Y_{12} + Y_{22} + Y_{32} = 0$
* $Y_{13} + Y_{23} + Y_{33} = 0$
* $Y_{31} + Y_{32} + Y_{33} = 0$

Con esas cuatro ecuaciones puedo hallar todos los términos de la MAI. Quedando:

### Matriz T en Source Común

Para esto primero planteo las ecuaciones:

$$V1 = V2 A + (-I2) B$$
$$V1 = V2 C + (-I2) D$$

Y con la matriz:
$$
Y_{SC}
=
\begin{bmatrix}
sC_{gs}+sC_{gd} & -C_{gd}s \\
g_m-sC_{gd} &
\dfrac{1}{r_{ds}}+sC_{gd}+sC_{ds} &
\end{bmatrix}
$$

Despejando se obtiene que para pasar genéricamente de una matriz Y a una matriz T tengo que hacer:
$$Y->T = -\frac{1}{Y_{21}}.
\begin{bmatrix}
Y_{22} & 1 \\
\Delta Y & Y_{11}
\end{bmatrix}
$$
Siendo $\Delta Y$ el determinante de la matriz Y.
